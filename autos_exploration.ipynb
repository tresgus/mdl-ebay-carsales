# eBay Used Car Listings

## Introduction

This project cleans and explores a sample of 50,000 used car listings scraped from *eBay Kleinanzeigen*, the classifieds section of the German eBay website, in March and April 2016. The raw data is messy. Prices and odometer readings are stored as text, many categories are in German, and some values (such as cars registered in the year 1000) are impossible. The goals of this notebook are to:

* clean the data so it is ready for analysis
* explore the cleaned listings, starting with how prices compare across German car brands

The notebook uses `pandas` and `numpy`.

## Data Cleaning

### Cleaning Columns

```python
autos.columns
```

```python
column_names = ['date_crawled', 'name', 'seller', 'offer_type', 'price', 'abtest', 'vehicle_type', 'registration_year', 
                'gear_box', 'power_ps', 'model', 'odometer', 'registration_month', 'fuel_type', 'brand', 'unrepaired_damage', 'ad_created',
                'num_photos', 'postal_code', 'last_seen']
autos.columns = column_names

autos.head()
```

Taking a look at the initial dataframe, there are some things to note. Most importantly, we see that the data has some column names that are a bit clunky and can be renamed for clarity. We also noted that the column names are in camelCase instead of snake_case, which is the convention preferred in pandas. So we reformatted the column names to snake_case.

### Data Exploration for Cleaning
After loading the data and renaming the columns, we can now dive into the data to inspect data quality and clean it. The first two steps we perform are to:
* drop any text columns where most of the values are the same
* clean and convert any text columns that store numeric data

```python
autos.describe(include='all')
```

After looking at the description table, we can see several characteristics of the columns that can help us decide how to clean our dataframe. The `seller` and `offer_type` columns have nearly all entries as the same value. We can drop these columns. The `num_photos` column looks a bit strange with 50,000 entries, but only one unique value. We will need to investigate more. Lastly, it looks like there are some numeric columns with text in them: `odometer` and `price`.

```python
autos["num_photos"].value_counts()
```

The `num_photos` column has an entry for every row, but every entry is 0. We will delete this column and the other two columns mentioned previously.

```python
autos = autos.drop(["num_photos", "seller", "offer_type"], axis=1)
```

We will clean the two columns with values stored as text: `odometer` and `price`. We will strip the non-numeric characters and convert them to the proper data types. We will also rename the `odometer` column to `odometer_km` to show that the readings are in kilometers, not miles, since the data source is scraped from a country where they use the metric system.

```python
autos["price"] = (autos["price"]
                    .str.replace("$", "")
                    .str.replace("," , "")
                    .astype(int)
                  )
autos["price"].head()
```

```python
autos["odometer"] = (autos["odometer"]
                        .str.replace("km", "")
                        .str.replace("kms", "")
                        .str.replace(',' , '')
                        .astype(int)
                     )

autos.rename({"odometer": "odometer_km"}, axis=1, inplace=True)
autos["odometer_km"].head()
```

### Dealing with Incorrect Registration Data

```python
autos.loc[autos["registration_year"] > 1930, "registration_year"].value_counts()
```

```python
autos.loc[autos["registration_year"] < 2016, "registration_year"].value_counts()
```

Realistically speaking, since the data was scraped in 2016, there should be no `registration_year` for values above 2016. Moreover, since cars were popularized around the 1930s, a good minimum value for the `registration_year` range is 1930. We want to double check the percentage of rows outside this range.

```python
(~autos["registration_year"].between(1930, 2016)).sum() / autos.shape[0]
```

Taking a look at the percentage of values that are not between 1930 and 2016, we can see that these are good parameters. Less than 4% of the data lies outside the upper and lower bounds.

```python
autos = autos[autos["registration_year"].between(1930, 2016)]
autos["registration_year"].value_counts(normalize=True)
```

Taking a look at the cleaned registration year, we can see that most of the cars were registered in recent years.

### Cleaning German Words

Several text columns still contain German words. Starting with `vehicle_type`, we will map each German category to its English equivalent. Categories that are already English (`bus`, `coupe`, `suv`) are left as they are.

```python
autos["vehicle_type"].value_counts(dropna=False)
```

```python
vehicle_type_english = {
    "limousine": "sedan",
    "kleinwagen": "small_car",
    "kombi": "station_wagon",
    "cabrio": "convertible",
    "andere": "other",
}

autos["vehicle_type"] = autos["vehicle_type"].replace(vehicle_type_english)
autos["vehicle_type"].value_counts(dropna=False)
```

Next we translate `gear_box`, `fuel_type`, and `unrepaired_damage` the same way. Values that are already English or are abbreviations (`diesel`, `lpg`, `cng`, `hybrid`) are left as they are.

```python
gear_box_english = {
    "manuell": "manual",
    "automatik": "automatic",
}

fuel_type_english = {
    "benzin": "petrol",
    "elektro": "electric",
    "andere": "other",
}

unrepaired_damage_english = {
    "ja": "yes",
    "nein": "no",
}

autos["gear_box"] = autos["gear_box"].replace(gear_box_english)
autos["fuel_type"] = autos["fuel_type"].replace(fuel_type_english)
autos["unrepaired_damage"] = autos["unrepaired_damage"].replace(unrepaired_damage_english)

for col in ["gear_box", "fuel_type", "unrepaired_damage"]:
    print(autos[col].value_counts(dropna=False), "\n")
```

## Data Exploration

### Price by Brand

```python
autos["brand"].value_counts(normalize=True)
```

Taking a look at the percentages for the car brands in the list, we can see the largest car brands. The top 5 brands in the list are manufactured in Germany: `volkswagen`, `bmw`, `opel`, `mercedes_benz`, and `audi`. The other German brands are `smart`, `porsche`, and `trabant`. We could investigate the aftermarket performance for German car brands by looking at the price ratio.

```python
brands_list = ["volkswagen", "bmw", "opel", "mercedes_benz", 
          "audi", "smart", "porsche", "trabant"]
german_brands = autos[autos["brand"].isin(brands_list)]
print(german_brands["brand"].unique())
```

```python
brand_mean_prices = {}

for brand in brands_list:
    brand_only = german_brands[german_brands["brand"] == brand]
    mean_price = brand_only["price"].mean()
    brand_mean_prices[brand] = int(mean_price)

print(brand_mean_prices)
```

Looking at the mean prices, the German brands fall into three clear groups:
* **Luxury:** `porsche` is in a league of its own, with a mean price of about `$45,600`, roughly five times more than any other German brand.
* **Premium:** `audi`, `mercedes_benz`, and `bmw` are close together, between about `$8,300` and `$9,300`.
* **Budget:** `volkswagen` sits in the middle at about `$5,400`, while `smart`, `opel`, and `trabant` are the cheapest at under `$3,600`.

## Conclusion

In this project we cleaned and explored 50,000 used car listings from eBay Kleinanzeigen. The raw data needed a lot of work before it could be analyzed:
* Column names were converted from camelCase to snake_case.
* The `seller`, `offer_type`, and `num_photos` columns were dropped because almost every row had the same value.
* `price` and `odometer` were stored as text, so we stripped the `$`, `km`, and commas and converted them to integers.
* Listings with unrealistic prices (`$0` or over `$350,000`) and impossible registration years (before 1930 or after 2016) were removed.
* German category values in `vehicle_type`, `gear_box`, `fuel_type`, and `unrepaired_damage` were translated to English.

After cleaning, we kept 46,674 listings, about 93% of the original data.

From exploring the cleaned data, we found that:
* Listings were crawled fairly evenly from early March to early April 2016, and most ads were created in the same window.
* Most cars have high mileage, with nearly two thirds of listings at the 150,000 km maximum.
* German brands dominate the site. The eight German brands make up about 63% of all listings, and `volkswagen` alone accounts for about 21%.
* Among German brands, price follows reputation. `porsche` is by far the most expensive, `audi`, `mercedes_benz`, and `bmw` form a premium tier, and `opel` and `trabant` are the cheapest.

One limitation is that we compared **mean** prices, which can be pulled up by a few very expensive listings, and some brands (such as `trabant`, with only 64 listings) have small samples. Next steps could include comparing median prices, checking whether mileage explains the price gaps between brands, and looking at how unrepaired damage affects price.
